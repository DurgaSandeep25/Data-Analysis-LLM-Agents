### Overview
In this notebook, we will be looking at sample outputs of PandasAgentWithTools. Main objective is to make the agent access the external real-world information (RAG system). Following we demonstrate how we can access each tool based on the given question
1. Weather API -  to access current weather condition
2. Wikipedia API - to access facts
3. Google Search - to access general information

This way, we can make our agents even more powerful by giving access to real-world information

```python
# import required packages
import sys
sys.path.append(".././part3")
from agent_with_tools import PandasAgentWithTools
import pandas as pd
```

```python
df = pd.read_csv(".././data/sample_data.csv")
print(df.shape)
df.head(1)
```

```python
tool_agent = PandasAgentWithTools(df=df)
```

```python
# uses weather tool
question = "Which state has the largest number of wildfire and what current weather in that State?"
print(tool_agent.run(question))
```

```python
# uses wikipedia to get the state's largest city + weather tool

question = "Which state has the smallest fire size, and what is the weather of that state's largest city for today?"
print(tool_agent.run(question))
```

```python
# uses google search for top-5 visits

question = "Which state has the largest number of wildfire and what are the top-5 places to visit in that State?"
print(tool_agent.run(question))
```

```python
# uses wikipedia tool for the facts about that state

question = "Which state has the largest fire size and tell me about some facts about that state?"
print(tool_agent.run(question))
```
